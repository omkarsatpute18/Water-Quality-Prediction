```python
df = df.drop_duplicates()
print("Duplicate rows successfully droped from the original dataset....",df.shape)
```

```python
X = df.drop("Potability",axis=1)
y = df["Potability"]
```

```python
print("Features shape:",X.shape)
print("Target shape:",y.shape)
```

```python
X.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
    X,y,random_state=42,test_size=0.20,stratify=y
)
```

```python
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
#We need to replace the nan values with some values like here i have replaced it with median value of it
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)

X_test_imputed = imputer.transform(X_test)
```

```python
#Checking the values for nan values
np.isnan(X_train_imputed).sum()
np.isnan(X_test_imputed).sum()
```

```python
#standard scaler calculates the mean and standard deviation of each numerical column and appplies to it

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

```python
print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)

print("Missing values in training data:", np.isnan(X_train_scaled).sum())
print("Missing values in testing data:", np.isnan(X_test_scaled).sum())
```

```python
#importing the LogisticRegression model cause it is classification problem it has 0 or 1
from sklearn.linear_model import LogisticRegression
```

```python
#Creating the Model 
model = LogisticRegression(random_state=42,class_weight="balanced")
```

```python
#Training on model using .fit
model.fit(X_train_scaled,y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
#First 20 predicted values
print("First 20 predictions:",y_pred[:20])
```

```python
#Actual vs predicted values by our LogisticRegression model
print("Actual values:")
print(y_test.values[:20])

print("\nPredicted values:")
print(y_pred[:20])
```

```python
#Checking the probability
y_prob = model.predict_proba(X_test_scaled)

print(y_prob[:5])
```

```python
#importing some importanat metrices which is needed to calculate the efficiency of model
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
#calculating accuracy_score
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
#precision_score
precision = precision_score(y_test, y_pred)

print("Precision:", precision)
```

```python
#recall score
recall = recall_score(y_test, y_pred)

print("Recall:", recall)
```

```python
#F1 score
f1 = f1_score(y_test, y_pred)

print("F1 Score:", f1)
```

```python
print("Model Evaluation Results")
print("------------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")
```

```python
#Classification report
print(classification_report(y_test,y_pred))
```

```python

#Confusion metrices
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
def predict_water_potability(
    ph,
    hardness,
    solids,
    chloramines,
    sulfate,
    conductivity,
    organic_carbon,
    trihalomethanes,
    turbidity
):
    
    # Create a DataFrame with the same feature order used during training
    new_data = pd.DataFrame({
        "ph": [ph],
        "Hardness": [hardness],
        "Solids": [solids],
        "Chloramines": [chloramines],
        "Sulfate": [sulfate],
        "Conductivity": [conductivity],
        "Organic_carbon": [organic_carbon],
        "Trihalomethanes": [trihalomethanes],
        "Turbidity": [turbidity]
    })
    
    # Apply the same preprocessing used during training
    new_data_imputed = imputer.transform(new_data)
    new_data_scaled = scaler.transform(new_data_imputed)
    
    # Make prediction
    prediction = model.predict(new_data_scaled)[0]
    
    # Get probability
    probability = model.predict_proba(new_data_scaled)[0]
    
    if prediction == 1:
        result = "Potable"
    else:
        result = "Not Potable"
    
    print("Water Potability Prediction")
    print("---------------------------")
    print("Prediction:", result)
    print(f"Probability of Not Potable: {probability[0] * 100:.2f}%")
    print(f"Probability of Potable    : {probability[1] * 100:.2f}%")
    
    return result
```

```python
predict_water_potability(
    ph=7.0,
    hardness=200,
    solids=20000,
    chloramines=7,
    sulfate=300,
    conductivity=400,
    organic_carbon=14,
    trihalomethanes=70,
    turbidity=4
)
```
